## Traffic Light System

### Goal:
    Simulate traffic light behavior.

### Task:

    Ask user to input current light color (red, yellow, or green).
    
    Use nested conditions to print what the driver should do.

### Example:

    Enter light color: red
    Stop! Wait for green light.

```python
while True:
    color = input('Enter light color (red/yellow/green): ').strip().lower()

    while color not in ['red', 'yellow', 'green']:
        print('Invalid light color. Try again.')
        color = input('Enter light color (red/yellow/green): ').strip().lower()

    if color == 'red':
        print('Stop! Wait for green light.')
    elif color == 'yellow':
        print('Wait and prepare to stop.')
    elif color == 'green':
        print('Go when the road is clear.')

    choice = input('Do you want to check another light? (yes/no): ').strip().lower()
    if choice != 'yes':
        break
```

## Course Registration System

### Goal:
    Allow students to register for available courses.

### Steps:

    courses = ["AI", "Python", "Math", "Physics"]
    
    Loop for multiple registrations.
    
    If course not in list → print “Not available” and continue
    
    If user enters “exit” → break
    
    Show registered courses.

### Example:

    Enter course: Python  
    Enter course: Biology → Not available  
    Enter course: AI  
    Enter course: exit  
    Registered: ['Python', 'AI']


```python
courses = ["AI", "Python", "Math", "Physics"]
registered=[]
while True:
    course = input('Enter course (or exit): ').strip()
    if course.lower() == 'exit':
        break
    match = next((c for c in courses if c.lower() == course.lower()), None)
    if match is None:
        print('Not available.')
        continue
    registered.append(match)
    print(match, 'registered.')
print('Registered:', registered)
```

## Create a menu-driven banking system using loops:
* Create Account
* Deposit
* Withdraw
* Check Balance
* Exit

```python
balance = 0.0
while True:
    print('\n1. Create Account\n2. Deposit\n3. Withdraw\n4. Check Balance\n5. Exit')
    choice = input('Choose an option: ')
    if choice == '1':
        print('Account is ready to use.')
    elif choice == '2':
        amount = float(input('Deposit amount: '))
        if amount > 0:
            balance += amount
        else:
            print('Amount must be positive.')
    elif choice == '3':
        amount = float(input('Withdraw amount: '))
        if 0 < amount <= balance:
            balance -= amount
        else:
            print('Invalid amount or insufficient balance.')
    elif choice == '4':
        print('Balance:', balance)
    elif choice == '5':
        print('Goodbye!')
        break
    else:
        print('Invalid choice.')
```

## Shop Sales Tracker
### Goal:

    Track daily sales of 3 products over 5 days and calculate the total revenue.

### Steps:

    Outer loop → iterate over 5 days.
    
    Inner loop → take sales amount for 3 products each day.
    
    Store daily total in a list.
    
    If any product sale = 0 → continue (skip that product).
    
    Print all daily totals and grand total.

### Example:
    Day 1:
    Product 1: 200
    Product 2: 150
    Product 3: 0 → skipped
    Day 2:
    Product 1: 300
    Product 2: 250
    Product 3: 400
    → Daily Totals = [350, 950], Grand Total = 1300

```python

daily_totals = []

for day in range(1, 6):
    print(f"\nDay {day}")
    daily_total = 0

    for product in range(1, 4):
        sale = int(input(f"Product {product}: "))

        if sale == 0:
            continue

        daily_total += sale

    daily_totals.append(daily_total)

grand_total = sum(daily_totals)

print("\nDaily Totals =", daily_totals)
print("Grand Total =", grand_total)
```

## Student Attendance Tracker
### Goal:

    Record attendance (Present/Absent) for multiple classes and students.

### Steps:

    Outer loop → each class (2 classes).
    
    Inner loop → each student (3 students).
    
    Store "P" or "A" in a nested list.

    Print full attendance chart.

### Example:
    Class 1:
    Student 1 (P/A): P
    Student 2 (P/A): A
    Student 3 (P/A): P
    Class 2:
    Student 1 (P/A): P

## Student Result Sheet
### Goal:

    Generate report cards for each student in a class.

### Steps:

    Outer loop → 5 students.
    
    Inner loop → 4 subjects.

    If marks < 33 → print “Failed in Subject”.

    Calculate total and grade per student.

### Example:
    Student 1: [80, 90, 70, 85] → Grade: A  
    Student 2: [50, 32, 45, 40] → Failed in Subject 2

```python
for student in range(1, 6):
    total = 0
    failed_subjects = []
    marks = []
    print(f'Student {student}:')
    for subject in range(1, 5):
        mark = float(input(f'Subject {subject} marks: '))
        marks.append(mark)
        total += mark
        if mark < 33:
            failed_subjects.append(subject)
    average = total / 4
    if failed_subjects:
        print('Failed in subject(s):', failed_subjects)
    if average >= 80:
        grade = 'A'
    elif average >= 70:
        grade = 'B'
    elif average >= 60:
        grade = 'C'
    elif average >= 50:
        grade = 'D'
    else:
        grade = 'F'
    print('Marks:', marks, 'Total:', total, 'Average:', average, 'Grade:', grade)
```

## Holiday Package Evaluator
### Goal:

Compare multiple travel packages across features like hotel, transport, food.

### Steps:

Outer loop → 3 packages.

Inner loop → 3 features.

Input rating out of 5.

Skip if rating = 0.

Print average rating per package.

### Example:
Package 1: [4, 5, 5] → Avg = 4.67  
Package 2: [3, 4, 0] → Avg = 3.5

```python
for package in range(1, 4):
    total = 0
    count = 0
    print(f'Package {package}: rate hotel, transport, and food from 1 to 5 (0 to skip).')
    for feature in ['Hotel', 'Transport', 'Food']:
        rating = float(input(f'{feature} rating: '))
        if rating == 0:
            continue
        if 1 <= rating <= 5:
            total += rating
            count += 1
        else:
            print('Invalid rating, skipped.')
    if count:
        print(f'Package {package} average rating: {total / count:.2f}')
    else:
        print(f'Package {package}: no ratings entered.')
```
